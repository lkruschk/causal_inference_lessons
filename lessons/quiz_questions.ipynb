__Question 1__

Consider the preemptive call program we discussed in lesson 2. As a reminder, in this example we imagined NRG implementing a call program to reach out to customers who we think will experience "bill shock" from unexpectedly high bills. 

Using this example, suppose there are two types of customers: those that experience bill shock ($S$) and those who do not ($N$). Further, suppose that our call program extends the tenure of each customer of type $S$ by 5 months, but has no effect on customers of type $N$ (customers experiencing bill shock are grateful for the call, whereas customers who don't experience bill shock don't care). Lastly, suppose that 10% of customers are type $S$ while 90% of customers are type $N$.

Now suppose we design an A/B test in which we randomly called 1000 customers, and compare their tenure to 1000 randomly selected customers we did not call. By how much would we expect the test to indicate our call program extends customer tenure? What is this causal effect called? Assume away any complicating factors outside those given in this question (e.g., survival rates and length of time between A/B test administration and evaluation).

a. The A/B test would indicate that our call program extends customer tenure by 5 months. This causal effect is called the average treatment effect on the treated (ATT).
</br>
b. The A/B test would indicate that our call program extends customer tenure by 2.5 months. This causal effect is called the average treatment effect on the treated (ATT).
</br>
c. The A/B test would indicate that our call program extends customer tenure by 0.5 months. This causal effect is called the average treatment effect (ATE).
</br>
d. The A/B test would indicate that our call program extends customer tenure by 0 months. This causal effect is called the average treatment effect (ATE).
</br>

__Question 2__

Recall the following relationship between observed outcomes and causal effects:
\begin{equation}
    \underbrace{E[Y_{1i}\text{ } | \text{ } D_{i} = 1] - E[Y_{0i} \text{ } | \text{ } D_{i} = 0]}_\text{Observed difference in group outcomes} = \underbrace{E[Y_{1i} - Y_{0i} \text{ } | \text{ } D_{i} = 1]}_\text{Average treatment effect on the treated} + \underbrace{E[Y_{0i}\text{ } | \text{ } D_{i} = 1] - E[Y_{0i} \text{ } | \text{ } D_{i} = 0]}_\text{Selection bias}
\end{equation}

Consider an A/B test in which we are evaluating a new online customer experience. Specifically, we randomly assign customers a "personalized experience", in which we suggest products to them based on their property's predicted usage. To test the efficacy of this program we measure customer conversion rate in both the test and control groups (e.g., $E[Y_{1i} \text{ }| \text{ } D_{i} = 1]$ is the observed average conversion rate for the treatment group).

Of course, not all products have the same conversion rate. This means that sometimes we recommend customers in the test group a product with a lower average conversion rate than what their expected conversion rate would have been in the control group. 

Based on the way this A/B is constructed, have we eliminated selection bias? If not, what sign does it take?

a. We have eliminated selection bias by randomly assigning customers to the test and control groups.
</br>
b. We have not eliminated selection bias because our test non-randomly assigns customers to different products. Selection bias is negative.
</br>
c. We have not eliminated selection bias because our test non-randomly assigns customers to different products. Selection bias is positive.
</br>
d. There is not enough information to answer this question.
